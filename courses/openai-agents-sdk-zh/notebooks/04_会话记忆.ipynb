## 第 1 步：创建 Agent 与本地会话

先预测这一步会做什么，再执行代码、观察返回结果。


```python
from agents import Agent, Runner, SQLiteSession\nfrom agent_course.config import model_name, output_dir, require_openai_key\nrequire_openai_key()\nagent = Agent(name='学习陪练', model=model_name(), instructions='使用中文简洁解释并参考前文。')\nsession = SQLiteSession('notebook-course', str(output_dir() / 'notebook_memory.sqlite3'))
```

## 第 2 步：两轮对话

先预测这一步会做什么，再执行代码、观察返回结果。


```python
r1 = await Runner.run(agent, '记住：我们正在学习肠道宏基因组测序。请用两句话介绍。', session=session)\nprint(r1.final_output)\nr2 = await Runner.run(agent, '现在基于刚才的话题给我一道问题。', session=session)\nprint(r2.final_output)
```
