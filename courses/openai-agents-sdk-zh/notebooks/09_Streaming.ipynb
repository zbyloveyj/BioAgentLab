# 流式输出事件

**完整中文教程｜配套脚本：** `lessons/09_streaming.py`  
**运行条件：** OpenAI API  

流式输出意味着随着模型生成，程序逐步显示文本，而不是等全部生成完才展示。它不代表模型推理过程可以被完整读取。

## 第 1 步：获取原始文本增量事件

先预测这一步会做什么，再执行代码、观察返回结果。


```python
from agents import Agent, Runner\nfrom openai.types.responses import ResponseTextDeltaEvent\nfrom agent_course.config import model_name, require_openai_key\nrequire_openai_key()\nagent = Agent(name='流式助理', model=model_name(), instructions='准确地用中文解释AI Agent')\nresult = Runner.run_streamed(agent, input='什么是工具调用？')\nasync for event in result.stream_events():\n    if event.type == 'raw_response_event' and isinstance(event.data, ResponseTextDeltaEvent):\n        print(event.data.delta, end='', flush=True)\nprint()
```
