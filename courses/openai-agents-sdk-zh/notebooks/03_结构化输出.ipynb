## 第 1 步：定义数据结构

先预测这一步会做什么，再执行代码、观察返回结果。


```python
from pydantic import BaseModel, Field\nfrom agents import Agent, Runner\nfrom agent_course.config import model_name, require_openai_key\nrequire_openai_key()\nclass Plan(BaseModel):\n    question: str\n    potential_confounders: list[str] = Field(description='潜在混杂因素')\n    next_step: str
```

## 第 2 步：请求结构化输出

先预测这一步会做什么，再执行代码、观察返回结果。


```python
agent = Agent(name='研究设计助理', model=model_name(), output_type=Plan,\n    instructions='只做研究规划，不编造具体结果。')\nresult = await Runner.run(agent, '构思一个宏基因组跨疾病比较研究。')\nprint(result.final_output.model_dump_json(indent=2))
```
