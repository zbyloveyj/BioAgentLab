## 第 1 步：创建专家并注册到总控

```python
from agents import Agent, Runner\nfrom agent_course.config import model_name, require_openai_key\nrequire_openai_key()\nmethod = Agent(name='方法顾问', model=model_name(), instructions='提出稳健研究设计，不捏造结果')\nwriter = Agent(name='写作顾问', model=model_name(), instructions='润色研究问题，不增加虚构事实')\nmanager = Agent(name='科研总控', model=model_name(),\n    instructions='先分别咨询两位专家，再给一个简短的研究方案。',\n    tools=[method.as_tool(tool_name='methods_advisor', tool_description='方法学建议'),\n           writer.as_tool(tool_name='writing_advisor', tool_description='中文科学表达')])
```
