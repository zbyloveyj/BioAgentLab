```python
from agents import Agent, Runner\nfrom agent_course.config import model_name, require_openai_key\nrequire_openai_key()\nagent = Agent(name='文献审读助手', model=model_name(),\n    instructions='只根据给定论文摘要回答；不可编造PMID，不可把相关性视为因果。')\nprompt = '\\n'.join(f\"PMID={r['pmid']} 标题={r['title']} 摘要={r['abstract'][:3000]}\" for r in records)\nif records:\n    result = await Runner.run(agent, prompt + '\\n请按 PMID 简要概述。')\n    print(result.final_output)\nelse:\n    print('当前检索无结果，跳过模型调用。')
```
